# Bivariate: Categorical + Categorical

```python
# Cross-tabulation: Country and Industry
country_industry_table = pd.crosstab(df['Country'], df['Industry'])
country_industry_table
```

```python
# Simple bar plot for the top countries and top industries
top_countries = df['Country'].value_counts().head(5).index
top_industries = df['Industry'].value_counts().head(5).index
country_industry_small = pd.crosstab(df['Country'], df['Industry']).loc[top_countries, top_industries]
country_industry_small.plot(kind='bar')
```

```python
import pandas as pd
import numpy as np

# Load the real Credit Card Default dataset directly from the public ISLR repository
url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/ISLR/Default.csv"
df = pd.read_csv(url)
```

```python
df
```

```python
# Convert 'default' to numeric (1 for Yes, 0 for No) so we can calculate percentages
df['default_rate'] = np.where(df['default'] == 'Yes', 1, 0)
```

```python
pd.crosstab(
    index=df['student'],
    columns='default_rate_mean',
    values=df['default_rate'],
    aggfunc='mean'
) * 100
```

```python
# Simple bar plot: default rate by student status
default_by_student = pd.crosstab(
    index=df['student'],
    columns='default_rate_mean',
    values=df['default_rate'],
    aggfunc='mean'
) * 100
default_by_student.plot(kind='bar')
```

```python
pd.crosstab(df['student'], df['default'], normalize='index') * 100
```

```python
df['balance_bucket'] = pd.cut(df['balance'],
                              bins=[0, 1000, 2000, 3000],
                              labels=['Low Debt ($0-$1k)', 'Medium Debt ($1k-$2k)', 'High Debt ($2k+)'])
```

```python
pd.crosstab(
    df['balance_bucket'],
    df['student'],
    values=df['default_rate'],
    aggfunc='mean'
) * 100
```

```python
# Simple bar plot: default rate by debt level and student status
default_by_balance_student = pd.crosstab(
    df['balance_bucket'],
    df['student'],
    values=df['default_rate'],
    aggfunc='mean'
) * 100
default_by_balance_student.plot(kind='bar')
```
