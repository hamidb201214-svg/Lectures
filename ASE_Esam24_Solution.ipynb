```python
# Husk at installere pakkerne, som du skal bruge
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Indlæs dataset fra github

data = pd.read_excel('https://raw.githubusercontent.com/aaubs/AnvendtStatistik2026/main/Data/data_eksamen24.xlsx')
```

```python
#Check at datasættet er tilgængeligt
print(data.shape)
data.head()
```

```python
# Univariate
# 2.1 - Job Satisfaction and Happiness
print("--- Frequency Tables ---")
data['SATJOB'].value_counts(normalize=True)
```

```python
# Univariate - Happiness
data['HAPPY'].value_counts(normalize=True)
```

```python
# Bivariate - Cross-tabulation table
pd.crosstab(data['SATJOB'], data['HAPPY'], normalize='index')
```

```python
# Bivariate - Job Satisfaction and Happiness
pd.crosstab(data['SATJOB'], data['HAPPY'], normalize='index').plot(kind='bar')
```

```python
# Bivariate - Job Satisfaction and Income
data.groupby('SATJOB')['INCOME'].describe()
```

```python
# Bivariate - Age and Income
data[['AGE', 'INCOME']].describe().T
```

```python
# Bivariate - Age and Income
data['AGE'].corr(data['INCOME'])
```

```python
# Bivariate - Age and Income
data[['AGE','INCOME']].corr()
```

```python
# Univariate - Age
data[['AGE']].plot(kind='hist', bins=40, figsize=(10, 4))
```

```python
# Univariate - Age and Income
data[['AGE', 'INCOME']].hist(bins=40, figsize=(10, 4))
```

```python
# Bivariate - Age and Income
data[['AGE', 'INCOME']].plot(kind='scatter', x='AGE', y='INCOME', figsize=(10, 4))
```

```python
# Bivariate - Happiness and Income
data.plot(kind='box', column='INCOME', by='HAPPY')
```

```python
# Bivariate - Happiness and Income
data.boxplot(column='INCOME', by='HAPPY')
```

```python
# Bivariate
data[['AGE', 'INCOME', 'SATJOB', 'HAPPY']].corr().round(2)
```

```python
import seaborn as sns
# Bivariate
sns.heatmap(data[['AGE', 'INCOME', 'SATJOB', 'HAPPY']].corr().round(2), annot=True, cmap='coolwarm')
```
